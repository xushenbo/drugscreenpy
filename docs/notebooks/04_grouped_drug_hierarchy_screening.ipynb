# Grouped Drug Hierarchy Screening

This notebook screens prescriptions after mapping product codes to BNF hierarchy labels.

```python
import pandas as pd
import drugscreenpy as eds
```

```python
therapy = pd.DataFrame(
    {
        "patid": [1, 2, 3],
        "prodcode": [10, 10, 11],
        "eventdate": ["2020-01-10"] * 3,
        "drugsubstance": ["drug_a"] * 3,
        "duration": [5, 5, 5],
    }
)
grouping = pd.DataFrame(
    {
        "prodcode": [10, 11],
        "bnf.section": ["section_1", "section_1"],
        "bnf.paragraph": ["paragraph_1", "paragraph_1"],
    }
)
eds.tl.assign_grouping_labels(therapy, level="section", mapping=grouping, output_col="drug")
```

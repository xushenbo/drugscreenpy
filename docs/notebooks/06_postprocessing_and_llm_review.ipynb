# Postprocessing and LLM Review

This notebook shows ranking helpers and the optional LLM review interface. The package accepts an injected callable; it does not require a specific LLM backend.

```python
import pandas as pd
import drugscreenpy as eds
```

```python
results = pd.DataFrame(
    {
        "drug": ["drug_a"],
        "disease": ["disease_x"],
        "N": [100],
        "Y": [30],
        "IRR": [2.5],
        "p.value": [0.001],
    }
)
eds.tl.rank_safety_hits(results, min_unexposed_events=1, min_exposed_events=1)
```

```python
def fake_llm(prompt: str) -> str:
    """Return a deterministic positive review for the tutorial."""
    return "yes"


pairs = pd.DataFrame({"drug": ["drug_a"], "disease": ["disease_x"]})
eds.tl.review_safety_symptoms(pairs, fake_llm)
```

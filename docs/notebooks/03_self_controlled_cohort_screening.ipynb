# Self-Controlled Cohort Screening

This notebook runs the basic substance-level self-controlled cohort workflow.

```python
import pandas as pd
import drugscreenpy as eds
```

```python
therapy = pd.DataFrame(
    {
        "patid": [1, 2, 3],
        "eventdate": ["2020-01-10"] * 3,
        "drugsubstance": ["drug_a"] * 3,
        "duration": [5, 5, 5],
    }
)
patients = pd.DataFrame(
    {
        "patid": [1, 2, 3],
        "dob": ["1970-01-01"] * 3,
        "frd": ["2019-01-01"] * 3,
        "tod": ["2021-01-01"] * 3,
        "lcd": ["2021-01-01"] * 3,
        "deathdate": ["2021-01-01"] * 3,
    }
)
events = pd.DataFrame(
    {
        "patid": [1, 2, 3],
        "disease": ["disease_x"] * 3,
        "disease_eventdate": pd.to_datetime(["2020-01-12", "2020-01-06", "2020-01-13"]),
    }
)
result = eds.tl.screen_substance_therapy(therapy, patients, events, workflow="actual", min_total_events=2)
result[["drug", "disease", "age.group", "IRR", "p.value"]]
```

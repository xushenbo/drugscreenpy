# Exposure Episode Construction

This notebook shows how raw therapy rows become exposure episodes.

```python
import pandas as pd
import ehrapy_drug_screening as eds
```

```python
therapy = pd.DataFrame({
    "patid": [1, 1, 2],
    "eventdate": ["2020-01-01", "2020-01-05", "2020-02-01"],
    "drugsubstance": ["drug_a", "drug_a", "drug_b"],
    "duration": [4, 4, 10],
})
patients = pd.DataFrame({"patid": [1, 2]})
prescriptions = eds.tl.prepare_prescriptions_from_therapy(therapy, patients=patients)
prescriptions
```

```python
episodes = eds.tl.build_exposure_episodes_from_prescriptions(prescriptions, gap_days=0, keep_first_only=False)
episodes
```
